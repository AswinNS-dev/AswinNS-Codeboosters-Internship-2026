```python
import pandas as pd

#  SAMPLE DATASET
data = {
    "name": [
        "Aarav Shah", "Priya Nair", "Rohan Mehta", "Sneha Rao", "Karthik Iyer",
        "Divya Pillai", "Arjun Kumar", "Meera Krishnan", "Vikram Das", "Ananya Reddy",
        "Suresh Babu", "Lakshmi Venkat", "Rahul Gupta", "Pooja Singh", "Arun Nair",
        "Deepa Thomas", "Nikhil Sharma", "Kavya Menon", "Sanjay Patel", "Riya Joshi"
    ],
    "gender": [
        "Male","Female","Male","Female","Male",
        "Female","Male","Female","Male","Female",
        "Male","Female","Male","Female","Male",
        "Female","Male","Female","Male","Female"
    ],
    "department": [
        "CS","CS","Electronics","Electronics","Mechanical",
        "CS","Mechanical","Electronics","CS","Mechanical",
        "Electronics","CS","Mechanical","Electronics","CS",
        "Mechanical","CS","Electronics","Mechanical","CS"
    ],
    "math_score": [88,92,74,85,60,95,70,83,91,77,65,89,72,86,93,68,80,76,55,97],
    "programming_score": [90,88,65,79,55,97,62,85,94,70,60,91,66,80,89,63,84,72,50,95],
    "attendance_percentage": [92,95,70,88,60,98,74,90,85,78,65,93,72,87,96,62,89,82,55,99],
    "total_score": [270,275,209,252,175,290,206,258,270,225,190,273,210,253,278,193,253,230,155,291]
}

df = pd.DataFrame(data)

SEP = "=" * 60
sep = "-" * 60

#  1. PROJECT TITLE
print(SEP)
print("       STUDENT PERFORMANCE EXPLORER")
print("         Powered by Pandas & Python")
print(SEP)

#  2. DATASET OVERVIEW
print("\n📋  DATASET OVERVIEW")
print(sep)
print(f"  Total Students        : {len(df)}")
print(f"  Total Columns         : {len(df.columns)}")
print(f"  Unique Departments    : {df['department'].nunique()}")

#  3. DEPARTMENT-WISE STUDENT COUNT
print("\n🏫  DEPARTMENT-WISE STUDENT COUNT")
print(sep)
dept_counts = df["department"].value_counts()
for dept, count in dept_counts.items():
    print(f"  {dept:<20} : {count} students")

#  4. SCORE ANALYSIS
print("\n📊  SCORE ANALYSIS")
print(sep)
print(f"  Highest Math Score        : {df['math_score'].max()}")
print(f"  Lowest  Math Score        : {df['math_score'].min()}")
print(f"  Avg Math Score            : {df['math_score'].mean():.2f}")
print(f"  Avg Programming Score     : {df['programming_score'].mean():.2f}")
print(f"  Avg Attendance Percentage : {df['attendance_percentage'].mean():.2f}%")

# ─────────────────────────────────────────────
#  5. GENDER-WISE AVERAGE TOTAL SCORE
# ─────────────────────────────────────────────
print("\n⚧️   GENDER-WISE AVERAGE TOTAL SCORE")
print(sep)
gender_avg = df.groupby("gender")["total_score"].mean()
for gender, avg in gender_avg.items():
    print(f"  {gender:<10} : {avg:.2f}")

# ─────────────────────────────────────────────
#  6. TOP 5 PERFORMERS
# ─────────────────────────────────────────────
print("\n🏆  TOP 5 PERFORMERS")
print(sep)
print(f"  {'Rank':<6}{'Name':<20}{'Department':<15}{'Total Score'}")
print(f"  {'-'*4:<6}{'-'*18:<20}{'-'*12:<15}{'-'*11}")
top5 = df.sort_values("total_score", ascending=False).head(5).reset_index(drop=True)
for i, row in top5.iterrows():
    print(f"  {i+1:<6}{row['name']:<20}{row['department']:<15}{row['total_score']}")

# ─────────────────────────────────────────────
#  7. ATTENDANCE ANALYSIS (< 75%)
# ────────────────────────────────────────────
print("\n⚠️   ATTENDANCE ANALYSIS (Below 75%)")
print(sep)
low_att = df[df["attendance_percentage"] < 75]
if low_att.empty:
    print("  ✅ All students have attendance above 75%. Great job!")
else:
    print(f"  {'Name':<20}{'Department':<15}{'Attendance %'}")
    print(f"  {'-'*18:<20}{'-'*12:<15}{'-'*12}")
    for _, row in low_att.iterrows():
        print(f"  {row['name']:<20}{row['department']:<15}{row['attendance_percentage']}%")

print(f"\n{SEP}")
```
