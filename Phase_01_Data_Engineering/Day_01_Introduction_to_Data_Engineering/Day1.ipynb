```python
cm = df[df['city'].isin(['Chennai', 'Mumbai'])]
print(cm[['name', 'city', 'programming_score']])
```

```python
ch = df[(df['city'] == 'Chennai') | (df['city'] == 'Mumbai')]
print(ch[['name','city','programming_score']])
```

```python
low_attendance = df[df['attendance_percentage']<=75]
print(f"Number of students with low percentage:{len(low_attendance)}")
print(low_attendance[['name','department','attendance_percentage']])
```

```python
df['total_score'] = df['math_score'] + df['science_score'] + df['english_score'] + df['programming_score']
top_student = df.loc[df['total_score'].idxmax()]
print("Topper student:")
print(top_student[['name','total_score']])
least_student = df.loc[df['total_score'].idxmin()]
print("\nLeast student:")
print(least_student[['name','total_score']])
```
