```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')
```

```python
raw_df = pd.read_excel('sales_data.xlsx')
print(f'{raw_df.shape[0]} rows, {raw_df.shape[1]} columns')
raw_df.head(10)
```

```python
print('Missing values:')
print(raw_df.isnull().sum())
print(f'\nDuplicates: {raw_df.duplicated().sum()}')
print(f'\nTypes:\n{raw_df.dtypes}')
```

```python
df = raw_df.copy()
print(f'Working copy: {df.shape}')
```

```python
before = df.isnull().sum().sum()

df['customer_name'].fillna('Unknown', inplace=True)
df['product'].fillna('Unknown', inplace=True)
df['category'].fillna('Other', inplace=True)
df['quantity'].fillna(df['quantity'].median(), inplace=True)
df['city'].fillna('Unknown', inplace=True)

after = df.isnull().sum().sum()
print(f'Nulls values : Before {before} -> After {after}')
```

```python
before = len(df)
df.drop_duplicates(inplace=True)
print(f'Duplicate Rows: Before {before} -> After {len(df)} removed {before - len(df)} duplicates')
```

```python
df.columns = df.columns.str.lower().str.replace(' ', '_').str.replace('-', '_')
```

```python
df['total_amount'] = df['quantity'] * df['unit_price']
print('Amount calculated')
```

```python
before = len(df)
df = df[(df['quantity'] > 0) & (df['unit_price'] > 0)]
print(f'Rows: {before} -> {len(df)} (removed {before - len(df)})')
```

```python
print(f'Original: {len(raw_df)} rows')
print(f'Cleaned: {len(df)} rows')
print(f'Removed: {len(raw_df) - len(df)}')
print(f'Nulls: {df.isnull().sum().sum()}')
print(f'Dupes: {df.duplicated().sum()}')
```

```python
summary = df.groupby('category').agg(
    orders=('order_id', 'count'),
    total=('total_amount', 'sum'),
    avg=('total_amount', 'mean')
).round(2).sort_values('total', ascending=False)
print(summary)
```

```python
print(f'Orders: {len(df)}')
print(f'Revenue: Rs {df["total_amount"].sum():,.0f}')
print(f'Average: Rs {df["total_amount"].mean():,.0f}')
print(f'Median: Rs {df["total_amount"].median():,.0f}')
```

```python
df.to_excel('sales_data_cleaned.xlsx', index=False)
print(f'Exported: sales_data_cleaned.xlsx ({len(df)} rows)')
```
