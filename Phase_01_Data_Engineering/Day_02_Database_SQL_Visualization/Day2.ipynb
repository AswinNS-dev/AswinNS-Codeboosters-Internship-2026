```python
import sqlite3
#sqlite is a built-in python library ,allows to create,connect and query SQLite databases,no pip install is needed
import pandas as pd
import matplotlib.pyplot as plt #to visualize the data's into insights
import warnings #It helps to filter out the warnings that raises like which is main line
warnings.filterwarnings('ignore')
print(f"pandas version{pd.__version__}")
print(f"sqlite3 version{sqlite3.version}")
```

```python
df = pd.read_csv("/content/drive/MyDrive/Data_Engineering_Internship/student_performance.csv")
print(f"Dataset loaded: {df.shape[0]} students, {df.shape[1]} columns")
print(f"columns:{df.columns.tolist()}")
print("\n First 3 rows")
df.head(3)
```

```python
conn = sqlite3.connect('college.db')
cursor = conn.cursor()
df.to_sql('students', conn, if_exists='replace', index=False)
cursor.execute("SELECT COUNT(*) FROM students")
count = cursor.fetchone()[0]
print("Database created")
print(f"table students has {count} rows")
```

```python
# PRAGMA is a special sqlite command(not standard sql)
# table_info() shows the structre of the table

cursor.execute("PRAGMA table_info(students)")
colums_info = cursor.fetchall()
print("Table structure:")
print(f"{'Column Name':<25} {'Data Type':<12}")
print("-"*38)

for col in colums_info:
  print(f"{col[1]:<25} {col[2]:<12}")
```

```python
def run_query(sql,description=""):
  if description:
    print(f"\n{"="*55}")
    print(f"{description}")
    print(f"\n{"="*55}")

  result = pd.read_sql_query(sql,conn)
  print(result.to_string(index=False))
  return result
```

```python
query1 = """SELECT student_id,name,department,math_score,attendance_percentage FROM students DESC LIMIT 10; """
result1 = run_query(query1,"")
```

```python
query2 = """SELECT student_id,name,department,math_score,attendance_percentage FROM students ORDER BY student_id DESC LIMIT 5; """
result1 = run_query(query2,"")
```

```python
query3 = """SELECT student_id,name,department,math_score,attendance_percentage FROM studentS LIMIT 5 offset 25; """
result1 = run_query(query3,"")
```

```python
query4 = """SELECT student_id,name,department,programming_score FROM students WHERE programming_score > 50 and programming_score < 75; """
result1 = run_query(query4,"")
```

```python
query9 = """SELECT student_id, name, department, programming_score FROM students WHERE department IN ('Computer Science', 'Electronics', 'Mechanical') ORDER BY programming_score DESC LIMIT 1;"""
result1 = run_query(query9, "Student with the maximum programming score from Computer Science, Electronics, or Mechanical departments")
```

```python
quer5 = """SELECT student_id,name,department,programming_score FROM students WHERE programming_score BETWEEN 50 and 75; """
result1 = run_query(quer5,"")
```

```python
query6 = """SELECT name,math_score,science_score,programming_score,attendance_percentage FROM students where department = 'Computer Science' ORDER BY programming_score DESC;"""
result1 = run_query(query6,"")
```

```python
query7 = """select department,COUNT(*) from students GROUP BY department;"""
result1 = run_query(query7,"")
```

```python
dept_data = {
    'dept_code': ['CS','EC','ME','CE'],
    'dept_name':['Computer Science','Electronics','Mechanical','Civil'],
    'hod_name':['Dr.Sharma','Dr.Reddy','Dr.Patel','Dr.Kumar'],
    'established':[1985,1988,1972,1965],
    'intake':[60,60,60,60]
}
dept_df = pd.DataFrame(dept_data)
dept_df.to_sql('departments',conn,if_exists='replace',index=False)
print("Created")
print(dept_df.to_string(index=False))

dept_map = {
    'Computer Science' : 'CS',
    'Electronics' : 'EC',
    'Mechanical' : 'ME',
    'Civil' : 'CE'
}
df['department'] = df['department'].map(dept_map)
df.to_sql('students',conn,if_exists='replace',index=False)
```
