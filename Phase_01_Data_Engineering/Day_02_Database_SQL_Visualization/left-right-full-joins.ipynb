```python
import pandas as pd

def run_query(df1, df2, how, on_col, description):
    print(f"\n--- {description} ---")
    print(f"Performing a {how.upper()} MERGE on '{on_col}'")
    result = pd.merge(df1, df2, on=on_col, how=how, suffixes=('_emp', '_dept'))
    print("\nResult:")
    display(result)
    return result


df_employees = pd.DataFrame({
    'emp_id': [101, 102, 103, 104, 105],
    'name': ['John Doe', 'Jane Smith', 'Peter Jones', 'Alice Brown', 'Bob White'],
    'dept_id': [1, 2, 1, 3, None]
})

df_departments = pd.DataFrame({
    'dept_id': [1, 2, 3, 4],
    'dept_name': ['HR', 'Sales', 'Marketing', 'IT']
})

print("Employees DataFrame:")
display(df_employees)

print("\nDepartments DataFrame:")
display(df_departments)
```

```python
run_query(df_employees, df_departments, how='left', on_col='dept_id', description="Left Join: All Employees and their Departments")
```

```python
run_query(df_employees, df_departments, how='right', on_col='dept_id', description="Right Join: All Departments and their Employees")
```

```python
run_query(df_employees, df_departments, how='outer', on_col='dept_id', description="Full Join: All Employees and All Departments")
```
