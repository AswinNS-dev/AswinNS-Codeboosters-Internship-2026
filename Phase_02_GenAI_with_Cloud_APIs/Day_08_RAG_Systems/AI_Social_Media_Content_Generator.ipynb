```python
import os
from langchain_groq import ChatGroq
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
os.environ["GROQ_API_KEY"] = input("Enter Your Groq API Key: ")
llm = ChatGroq(model="llama-3.3-70b-versatile", temperature=0)
```

```python

linkedin_prompt = PromptTemplate(
    input_variables=["topic", "word_count"],
    template="""Generate a professional LinkedIn post about: {topic} with the Requirements:
Strong hook in the first line, Professional tone with industry insights,Clear Call-To-Action at the end,Exactly {word_count} words,3 to 5 relevant hashtags
"""
)

instagram_prompt = PromptTemplate(
    input_variables=["topic", "word_count"],
    template="""Generate an Instagram caption about: {topic}
Requirements:Use emojis throughout, High engagement and conversational tone, Ask a question to boost comments, Exactly {word_count} words, 10 to 15 relevant hashtags
"""
)

x_prompt = PromptTemplate(
    input_variables=["topic", "word_count"],
    template="""Generate a concise viral X (Twitter) post about: {topic} Requirements:Bold hook in the first sentence, Punchy and shareable style, Be mindful of 280-character limit, Exactly {word_count} words, 2 to 3 hashtags only
"""
)

facebook_prompt = PromptTemplate(
    input_variables=["topic", "word_count"],
    template="""Generate a Facebook post about: {topic} Requirements:Warm and conversational tone, Encourage readers to leave a comment, End with an open-ended question, Exactly {word_count} words, 3 to 5 hashtags
"""
)

threads_prompt = PromptTemplate(
    input_variables=["topic", "word_count"],
    template="""Generate a Threads post about: {topic} Requirements:Casual and authentic tone, Spark a discussion or debate, Feels like a genuine opinion, not an ad, Exactly {word_count} words, 1 to 2 hashtags only
"""
)

youtube_prompt = PromptTemplate(
    input_variables=["topic", "word_count"],
    template="""Generate a YouTube Community post about: {topic} Requirements:Address subscribers directly, Encourage likes, comments and interaction, Build excitement and a sense of community, Exactly {word_count} words, 2 to 3 hashtags
"""
)

platform_prompts = {
    "linkedin":          linkedin_prompt,
    "instagram":         instagram_prompt,
    "x":                 x_prompt,
    "facebook":          facebook_prompt,
    "threads":           threads_prompt,
    "youtube":           youtube_prompt,
}
```

```python
parser = StrOutputParser()

def generate_content(topic, platform, word_count):
    prompt = platform_prompts[platform.lower()]
    chain = prompt | llm | parser
    return chain.invoke({"topic": topic, "word_count": word_count})

def show_output(platform, content, variation):
    print("=" * 50)
    print(f"{platform.upper()} — Variation {variation}")
    print("=" * 50)
    print(content)
    print(f"\nWord Count : {len(content.split())}")
    print(f"Char Count : {len(content)}")
    print("=" * 50)
```

```python
supported = ["LinkedIn", "Instagram", "X (Twitter)", "Facebook", "Threads", "YouTube"]
print("Supported Platforms:", supported)

while True:
    topic = input("\nEnter topic (or 'exit'): ")
    if topic.lower() == "exit":
        break

    platform = input("Enter platform: ")
    if platform.lower() not in platform_prompts:
        print("Invalid platform. Choose from:", ", ".join(supported))
        continue

    word_count = input("Enter word count: ")

    print("\nGenerating 3 variations...\n")
    for i in range(1, 4):
        content = generate_content(topic, platform, word_count)
        show_output(platform, content, i)

    choice = input("\n[R] Regenerate     [Q] Quit: ").strip().lower()
    if choice == "q":
        break
    elif choice == "r":
        print("\nRegenerating...\n")
        for i in range(1, 4):
            content = generate_content(topic, platform, word_count)
            show_output(platform, content, i)
```
