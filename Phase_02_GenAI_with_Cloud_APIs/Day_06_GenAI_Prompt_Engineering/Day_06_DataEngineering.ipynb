```python
prompt=PromptTemplate(
    input_variables=["question"],
    template="""You are a student.

Rules:
1. Teacher speaks in English.
2. Student replies ONLY in Tamil.
3. Answer naturally and briefly.
4. Do not generate teacher questions.

Teacher: {question}
Student:


"""
)
parser=StrOutputParser()
chain=prompt |llm|parser

while True:
    teacher_question = input("Teacher: ")

    if teacher_question.lower() == "exit":
        break
    response=chain.invoke({
        "question":teacher_question
})
print(response)
```
